```python
from sklearn.model_selection import train_test_split

# Separate features and target
X = df.drop("isFraud", axis=1)
y = df["isFraud"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y
)

print("Original training distribution:")
print(y_train.value_counts())

print("\nOriginal test distribution:")
print(y_test.value_counts())
```

```python
# Combine training X and y temporarily
train_df = X_train.copy()
train_df["isFraud"] = y_train.values

# Separate classes
fraud = train_df[train_df["isFraud"] == 1]
normal = train_df[train_df["isFraud"] == 0]

print("Fraud:", len(fraud))
print("Normal:", len(normal))
```

```python
from sklearn.model_selection import train_test_split

# Split dataset into train and test
train, test = train_test_split(df, test_size=0.2, random_state=42)

balanced_train = train.copy()
```

```python
# Separate the classes
fraud = balanced_train[balanced_train["isFraud"] == 1]
normal = balanced_train[balanced_train["isFraud"] == 0]

# Take first len(fraud) normal records
normal_sample = normal.iloc[:len(fraud)]

# Combine both classes
balanced_50_50 = pd.concat([normal_sample, fraud], axis=0)

# Reset index
balanced_50_50 = balanced_50_50.reset_index(drop=True)

# Check distribution
print(balanced_50_50["isFraud"].value_counts())
```

```python
print("Original training rows:", len(X_train))
print("Balanced training rows:", len(balanced_50_50))
```

```python
important_columns = [
    "amount",
    "oldbalanceOrg",
    "newbalanceOrig",
    "oldbalanceDest",
    "newbalanceDest"
]

print("Original Training:")
print(X_train[important_columns].describe())

print("\nBalanced Training:")
print(balanced_50_50[important_columns].describe())
```

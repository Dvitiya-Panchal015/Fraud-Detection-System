```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/PS_20174392719_1491204439457_log.csv"
)

print("Dataset shape:", df.shape)
```

```python
X = df.drop("isFraud", axis=1)
y = df["isFraud"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# combine training data

train_data = X_train.copy()

train_data["isFraud"] = y_train.values

print(train_data.shape)
```

```python
# separate fraud and normal

fraud = train_data[train_data["isFraud"] == 1]

normal = train_data[train_data["isFraud"] == 0]

print("Fraud records:", len(fraud))
print("Normal records:", len(normal))
```

```python
# Non- random balancing 

normal_sample = normal.iloc[:len(fraud)]
```

```python
balanced_train = pd.concat(
    [normal_sample, fraud],
    axis=0
).reset_index(drop=True)
```

```python
# check the balance

print(balanced_train["isFraud"].value_counts())
```

```python
# create final training X and y

X_train_balanced = balanced_train.drop(
    "isFraud",
    axis=1
)

y_train_balanced = balanced_train["isFraud"]
```

```python
print("X_train_balanced:", X_train_balanced.shape)
print("y_train_balanced:", y_train_balanced.shape)
```
